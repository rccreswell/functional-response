```python
import pints
import numpy as np
import matplotlib.pyplot as plt
import pandas
```

```python
dfs = []
for i in range(3):
    dfs.append(pandas.read_csv('chains_all_h_chain{}.csv'.format(i)))

data = np.asarray(dfs)
```

```python
names = dfs[0].columns
names = [s.replace('_', '\_') for s in names]

summary = pints.MCMCSummary(data, parameter_names=names)

x = summary.summary()
```

```python
df = pandas.DataFrame(x, columns=['Parameter', 'Post.\ Mean', 'Post.\ Std.\ Dev.', '2.5 pctle', '25 pctle', '50 pctle', '75 pctle', '97.5 pctle', '$\hat{R}$', 'ess'])
df = df.drop(['ess'], axis=1)
df.to_latex('table1.tex', index=False, float_format='{:0.3f}'.format)
```
